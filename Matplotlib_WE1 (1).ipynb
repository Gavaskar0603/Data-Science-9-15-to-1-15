```python
import matplotlib.pyplot as plt
```

```python
df.groupby("grade")[["total_score","weekly_self_study_hours"]].mean().plot(kind="bar")
plt.title("Students performance")
plt.xlabel("Grade")
plt.ylabel("Total Score")
plt.legend()
plt.show()
```
