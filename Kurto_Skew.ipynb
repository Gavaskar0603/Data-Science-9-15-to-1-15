```python
import numpy as np
```

```python
norm_dist = np.random.normal(loc=50,scale=3,size=100) #value(mean),scale(std),size()
norm_dist
```

```python
np.mean(norm_dist)
```

```python
np.median(norm_dist)
```

```python
od = np.array([1000,1570,2500,3000,5000])
od
```

```python
na = np.concatenate([norm_dist,od])
na
```

```python
np.mean(na)
```

```python
np.median(na)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame({"Normal_Dist":norm_dist})
df
```

```python
df1 = pd.DataFrame({"Normal_Dist_OD":na})
df1
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10,7))
plt.subplot(1,2,1)
sns.histplot(norm_dist,kde=True)
plt.title("Normal Dist")
plt.subplot(1,2,2)
sns.histplot(na,kde=True)
plt.title("Outlires")
plt.show
```
