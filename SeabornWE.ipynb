```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
d=sns.load_dataset("iris")
d
```

```python
d1=sns.load_dataset("tips")
d1
```

```python
sns.scatterplot(x="total_bill",y="tip",data=d1)
plt.show()
```

```python
sns.histplot(x="size",data=d1)
plt.show()
```

```python
sns.scatterplot(x=d1["total_bill"],y=d1["tip"])
plt.show()
```

```python
d1.info()
```

```python
cr=d1.select_dtypes(["float64","int64"])
cr
```

```python
c=cr.corr()
c
```

```python
sns.heatmap(c,annot=True)
plt.show()
```

```python
sns.pairplot(d1,hue="sex")
plt.show()
```

```python
sns.pairplot(d,hue="species")
plt.show()
```

```python
f=sns.FacetGrid(d1,col="sex")
f.map(sns.scatterplot,"total_bill","tip")
plt.show()
```

```python
f=sns.FacetGrid(d,col="species")
f.map(sns.histplot,"sepal_length","sepal_width")
plt.show()
```
